```python
# Replicate exactly what the agent tool calls
def agent_search(query: str, hierarchy_filter: str = "", k: int = 10):
    results = search_snomed_concepts(
        query=query,
        resources=resources,
        hierarchy_filter=hierarchy_filter or None,
        k=k,
    )
    # for r in results:
    #     print(f"  {r['id']} | {r['label'][:55]:<55} | hier={r.get('top_level_hierarchy','')} | score={r.get('fused',0):.3f}")
    return results
```
